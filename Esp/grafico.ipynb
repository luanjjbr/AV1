```python
import json
import matplotlib.pyplot as plt

def plot_teleplot_dual(json_file, signal_x, signal_y, title="Entrada vs Saída", relative_time=True):

    with open(json_file, "r") as f:
        data = json.load(f)

    telemetries = data["telemetries"]

    if signal_x not in telemetries:
        raise ValueError(f"Sinal '{signal_x}' não encontrado.")
    if signal_y not in telemetries:
        raise ValueError(f"Sinal '{signal_y}' não encontrado.")

    sx = telemetries[signal_x]
    sy = telemetries[signal_y]

    tx = sx["data"][0]
    x = sx["data"][1]

    ty = sy["data"][0]
    y = sy["data"][1]

    if relative_time:
        t0 = min(tx[0], ty[0])
        tx = [t - t0 for t in tx]
        ty = [t - t0 for t in ty]

    fig, ax1 = plt.subplots()

    ax1.plot(tx, x, "b", label=signal_x)
    ax1.set_xlabel("Tempo (s)")
    ax1.set_ylabel(signal_x, color="b")
    ax1.tick_params(axis="y", labelcolor="b")
    ax1.grid(True)

    ax2 = ax1.twinx()
    ax2.plot(ty, y, "r", label=signal_y)
    ax2.set_ylabel(signal_y, color="r")
    ax2.tick_params(axis="y", labelcolor="r")

    plt.title(title)

    # ============================
    # Legenda no canto superior esquerdo
    # ============================
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()

    ax1.legend(
        lines1 + lines2,
        labels1 + labels2,
        loc="upper left"
    )

    plt.show()
```

```python
plot_teleplot_dual(
    "8_im.json",
    "Entrada",
    "Saida_Y",
    "Resposta do Sistema 8: Entrada vs Saída"
)
plot_teleplot_dual(
    "8_D.json",
    "Entrada",
    "Saida_Y",
    "Resposta do Sistema 8: Entrada vs Saída"
)
plot_teleplot_dual(
    "8_R.json",
    "Entrada",
    "Saida_Y",
    "Resposta do Sistema 8: Entrada vs Saída"
)
```
